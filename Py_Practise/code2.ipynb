```python
data = "he9llo wo7rld"
count = {"letter": 0, "space": 0, "number":0}

for char in data:
    if char.isalpha():
        count["letter"] += 1
    elif char == " ":
        count["space"] += 1
    else:
        count["number"] += 1

print(count)


# {'letter': 10, 'space': 1, 'number': 2}
```

```python
data = "hello world 35"
count = {}

for char in data:
    if char in count:
        count[char] += 1
    else:
        count[char] = 1
print(count)

# {'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 2, 'w': 1, 'r': 1, 'd': 1, '3': 1, '5': 1}
```

```python
def Palindrome_Check(s):
    rev = "" 
    for i in s:
        rev = i + rev
    if rev == s:
        return "Palidrome"
    else:
        return "Not Palidrome"
data = "12321"
Palindrome_Check(data)
```

```python
class Person:

    def __init__(self,name):
        self.name = name

    def display_name(self):
        print("Name: ", self.name)
class Employee(Person):

    def __init__(self, name, EmpID):
        super().__init__(name)
        self.EmpID = EmpID

    def EmpDetails(self):
        print("Name: ",self.name, "Employee_ID: ", self.EmpID)
        

obj = Person("Uday")
obj.display_name()


obj2 = Employee("Uday", "ID93871")
obj2.EmpDetails()


# Name:  Uday
# Name:  Uday Employee_ID:  ID93871
```

```python
# Encapsulation

class Bank:
    def __init__(self, balance):
        self.__balance = balance   # private variable

    def deposit(self, amt):
        self.__balance += amt

    def withdraw(self, amt):
        self.__balance -= amt

    def get_balance(self):
        return self.__balance


b = Bank(10000)
# b.deposit(1000)
b.withdraw(5000)
print(b.get_balance())

# 5000
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):

    def start(self):
        print("Car starts with key")

class Bike(Vehicle):

    def start(self):
        print("Bike starts without key")

obj = Car() 
obj.start()

obj2 = Bike()
obj2.start()

'''Car starts with key
Bike starts without key'''
```

```python
# Inheritance

class Animal:

    def speak(self):
        print("Animal makes sound")
class Dog(Animal):
    pass

obj = Dog()
obj.speak()

# Animal makes sound
```

```python
#Polymorphism


class Dog:

    def sound(self):
        print("Bark")

class Cat:

    def sound(self):
        print("Meow")


for animal in (Dog(), Cat()):
    animal.sound()

'''Bark
Meow'''
```

```python
#Multiple inheritnace

class A:
    def show(self):
        print("A")


class B:
    def display(self):
        print("B")


class C(A, B):
    pass

C().show()
C().display()

'''A
B'''
```

```python

#Multilevel

class A:
    def show(self):
        print("A")


class B(A):
    pass


class C(B):
    pass

C().show()

# A
```

```python
#Hierarchical

class A:

    def show(self):
        print("A")

class B(A):
    pass

class C(A):
    pass

B().show()
C().show()

'''A
A'''
```
